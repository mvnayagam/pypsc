# Example to solve structures within NEPA framework

```python
import logging
from psc.config.logconfig import setup_logging
# import uuid
import numpy as np

from psc.lib.pscsolver import PSCSolverNEPA

# ------ USAGE
solver = PSCSolverNEPA(reflections=np.arange(1,5), structurefactor=[10, 8, 1], structure=[0.328,0.218, 0.125], intensity_type='intensity', write_polytope=False)
result = solver.solve()
result
```
